```python
import pandas 
import numpy as np
import matplotlib.pyplot as plt
import glob

def parse_list(csv_raw_cell):
    csv_list_cell = list()
    for el in str(csv_raw_cell).split(","):
        csv_list_cell.append(float(el))
    return np.array(csv_list_cell)


stats_dir = "/home/dopf02-admin/Carla-CTS02_backup/output/is_despot_noise05/metrics/test/"
key = "S-10"
performance_dfs = []
csv_files = glob.glob(stats_dir + "*.csv")
for csv_file_path in csv_files:
    if key in csv_file_path: continue
    print(csv_file_path)
    with open(csv_file_path, "r") as file:
        file.readline()
        scenario_df = pandas.read_csv(file)
        print(scenario_df.shape)
        performance_dfs.append(scenario_df)

merged_df = performance_dfs[0]
for scenario_df in performance_dfs[1:]:
    merged_df = pandas.concat([merged_df, scenario_df], axis="index")
    print(merged_df.shape)
```
